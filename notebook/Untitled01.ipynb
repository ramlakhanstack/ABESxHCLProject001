```python
# Heart Failure Records - Exploratory Data Analysis (EDA)

# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


# 1. Load Dataset


url = "/content/heart+failure+clinical+records.zip"

df = pd.read_csv(url)

print("\n========== DATASET LOADED ==========")


# 2. Basic Data Exploration


print("\nFirst 5 rows:")
print(df.head())

print("\nLast 5 rows:")
print(df.tail())

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()

print("\nStatistical Description:")
print(df.describe())

# ----------------------------------------
# 3. Missing Values
# ----------------------------------------

print("\nMissing Values:")
print(df.isnull().sum())


# 4. Duplicate Records


print("\nDuplicate Records:")
print(df.duplicated().sum())


# 5. Age Distribution


print("\nAge Statistics:")
print("Mean Age:", df["age"].mean())
print("Median Age:", df["age"].median())
print("Minimum Age:", df["age"].min())
print("Maximum Age:", df["age"].max())

# NumPy array calculation
age_array = df["age"].to_numpy()

print("\nNumPy Age Analysis:")
print("Average Age using NumPy:", np.mean(age_array))
print("Standard Deviation:", np.std(age_array))


# 6. Death Event Analysis


print("\nDeath Event Count:")
print(df["death_event"].value_counts())

print("\nDeath Event Percentage:")
print(df["death_event"].value_counts(normalize=True) * 100)

# 7. Numerical Columns Analysis


columns = ["age", "ejection_fraction", "serum_creatinine"]

print("\nNumerical Data Analysis:")

for col in columns:
    print("\nColumn:", col)
    print("Mean:", df[col].mean())
    print("Median:", df[col].median())
    print("Minimum:", df[col].min())
    print("Maximum:", df[col].max())

# ----------------------------------------
# 8. Group Analysis
# ----------------------------------------

print("\nAverage values grouped by death_event:")

group_analysis = df.groupby("death_event")[
    ["ejection_fraction", "serum_creatinine"]
].mean()

print(group_analysis)

# ----------------------------------------
# 9. Diabetes, High BP and Anaemia
# ----------------------------------------

print("\nDiabetes Count:")
print(df["diabetes"].value_counts())

print("\nHigh Blood Pressure Count:")
print(df["high_blood_pressure"].value_counts())

print("\nAnaemia Count:")
print(df["anaemia"].value_counts())

# ----------------------------------------
# 10. Filtering Data
# ----------------------------------------

print("\nRecords with age greater than 60:")
print(df[df["age"] > 60].head())

# ----------------------------------------
# 11. Sorting Data
# ----------------------------------------

print("\nTop 10 records sorted by age:")
print(df.sort_values(by="age", ascending=False).head(10))

# ----------------------------------------
# 12. Death Event Bar Chart
# ----------------------------------------

death_counts = df["death_event"].value_counts().sort_index()

plt.figure(figsize=(7, 5))

plt.bar(
    death_counts.index.astype(str),
    death_counts.values
)

plt.title("Death Event Count")
plt.xlabel("Death Event (0 = No, 1 = Yes)")
plt.ylabel("Number of Records")

plt.tight_layout()
plt.show()

# ----------------------------------------
# 13. Age Histogram
# ----------------------------------------

plt.figure(figsize=(8, 5))

plt.hist(df["age"], bins=15, edgecolor="black")

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()

# ----------------------------------------
# 14. Final Summary
# ----------------------------------------

print("\n========== PROJECT COMPLETED ==========")
print("Total Records:", df.shape[0])
print("Total Columns:", df.shape[1])
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Records:", df.duplicated().sum())

print("\nEDA Completed Successfully!")
```
